```python
# PROVIDED HELPER — NO EDITS REQUIRED
# Uses sklearn permutation_importance with the fitted final model.
import numpy as np
import pandas as pd
from sklearn.inspection import permutation_importance
from sklearn.metrics import mean_absolute_error

class _FunctionRegressor:
    """Adapter exposing .predict(X) on original/raw feature columns."""
    def __init__(self, prediction_function):
        self.prediction_function = prediction_function

    def fit(self, X, y=None):
        raise RuntimeError("The model is already fitted; do not refit it.")

    def predict(self, X):
        predictions = np.asarray(self.prediction_function(X)).reshape(-1)
        if len(predictions) != len(X):
            raise ValueError("Prediction function must return one prediction per row")
        return predictions

def regression_permutation_importance(predict_fn, X_raw, y_true,
                                      n_repeats=5, random_state=42):
    """Return change in MAE from permuting each original feature.

    predict_fn(X_raw) must apply the fitted final preprocessing and model.
    X_raw must be a DataFrame of original predictor columns.
    Call using the finalized model and test data, for interpretation only.
    No fitting or changes to the model occur inside this function.
    """
    if not isinstance(X_raw, pd.DataFrame):
        raise TypeError("X_raw must be a DataFrame with original feature names")
    adapter = _FunctionRegressor(predict_fn)
    result = permutation_importance(
        adapter, X_raw, y_true,
        scoring=lambda estimator, X, y: -mean_absolute_error(y, estimator.predict(X)),
        n_repeats=n_repeats, random_state=random_state,
    )
    return pd.DataFrame({
        "feature": X_raw.columns,
        "importance_mean": result.importances_mean,
        "importance_std": result.importances_std,
    }).sort_values("importance_mean", ascending=False).reset_index(drop=True)
```

## Appendix — Understanding Regression Diagnostic Plots
The following code generates synthetic actual and predicted values, introduces some systematic prediction errors, and displays the three diagnostic plots required in Problem 4.

These are **synthetic example data**, not project test data. For Problem 4, Part C, use the actual and predicted test values that you saved in Part B. The plots help you interpret the finalized model; **do not use the test results to revise the model**.


```python

import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
n = 500

# Synthetic actual values, roughly resembling house prices
actual = rng.uniform(100_000, 800_000, n)

# Predictions with:
# - Some regression toward the mean
# - Increasing error variability for expensive properties
# - A few unusually large errors
noise = rng.normal(0, 15_000 + 0.08 * actual, n)

predicted = 450_000 + 0.85 * (actual - 450_000) + noise

# Introduce a few unusually large prediction errors
outliers = rng.choice(n, size=8, replace=False)
predicted[outliers] += rng.choice([-1, 1], size=8) * 180_000

# Residuals: positive means underprediction
residuals = actual - predicted

# --------------------------------------------------
# Plot 1: Actual vs. Predicted
# --------------------------------------------------
plt.figure(figsize=(7, 5))
plt.scatter(actual, predicted, alpha=0.4)

lo = min(actual.min(), predicted.min())
hi = max(actual.max(), predicted.max())

plt.plot([lo, hi], [lo, hi], "r--", label="Perfect prediction")
plt.xlabel("Actual value ($)")
plt.ylabel("Predicted value ($)")
plt.title("Actual vs. Predicted")
plt.legend()
plt.show()

# --------------------------------------------------
# Plot 2: Residuals vs. Predicted
# --------------------------------------------------
plt.figure(figsize=(7, 5))
plt.scatter(predicted, residuals, alpha=0.4)
plt.axhline(0, color="red", linestyle="--")

plt.xlabel("Predicted value ($)")
plt.ylabel("Residual: actual - predicted ($)")
plt.title("Residuals vs. Predicted")
plt.show()

# --------------------------------------------------
# Plot 3: Residual Distribution
# --------------------------------------------------
plt.figure(figsize=(7, 5))
plt.hist(residuals, bins=35, edgecolor="black", alpha=0.75)
plt.axvline(0, color="red", linestyle="--")

plt.xlabel("Residual ($)")
plt.ylabel("Frequency")
plt.title("Residual Distribution")
plt.show()
```
